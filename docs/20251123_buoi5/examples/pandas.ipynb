```python
import pandas as pd

# Series 1: Cars
cars = pd.Series(["VW", "Ford", "Toyota", "Nissan"], name="Cars")

# Series 2: In stock
in_stock = pd.Series([25, 4, 10, 0], name="In stock")

# Ghép 2 Series thành DataFrame
df = pd.concat([cars, in_stock], axis=1)

print("Series - Cars:")
print(cars)
print("\nSeries - In stock:")
print(in_stock)
print("\nDataFrame (Cars + In stock):")
print(df)
```
