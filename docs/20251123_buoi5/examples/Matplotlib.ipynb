```python
import matplotlib.pyplot as plt

# Dữ liệu
categories = ['A', 'B', 'C', 'D', 'E']
values = [23, 45, 12, 36, 27]

# Vẽ
fig, ax = plt.subplots()
bars = ax.bar(categories, values, width=0.6)

# Tùy biến
ax.set_title('Bar Graph Example')
ax.set_xlabel('Category')
ax.set_ylabel('Value')
ax.grid(axis='y', linestyle='--', alpha=0.4)

# (Tùy chọn) ghi giá trị trên đỉnh cột
for b in bars:
    ax.text(b.get_x() + b.get_width()/2, b.get_height(), f'{b.get_height()}',
            ha='center', va='bottom', fontsize=9)

plt.show()
```

```python
import matplotlib.pyplot as plt

labels = ['Apples', 'Bananas', 'Oranges', 'Grapes']
values = [30, 25, 20, 25]
explode = [0.05, 0, 0, 0]  # làm nổi bật lát đầu tiên

fig, ax = plt.subplots()
ax.pie(values, labels=labels, explode=explode,
       autopct='%1.1f%%', startangle=90, pctdistance=0.8)
ax.set_title('Pie Chart Example')
ax.axis('equal')  # giữ hình tròn

plt.show()
```

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [2.3, 2.7, 3.1, 3.6, 4.0, 4.5]

fig, ax = plt.subplots()
ax.scatter(x, y, marker='o')  # có thể đổi marker, ví dụ 's', '^', 'x', ...
ax.set_title('Scatter Plot (Matplotlib)')
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.grid(True, alpha=0.3)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)
x = np.linspace(0, 10, 50)
y = 0.5 * x + 1 + np.random.normal(scale=1.0, size=x.size)
z = np.linspace(0, 1, x.size)  # biến thứ ba để mã hóa màu/kích thước

fig, ax = plt.subplots()
sc = ax.scatter(x, y, c=z, s=50 + 200*z, alpha=0.85)  # c: màu theo z, s: kích thước theo z
cbar = plt.colorbar(sc, ax=ax)
cbar.set_label('z (magnitude)')

ax.set_title('Scatter with colormap & size mapping')
ax.set_xlabel('X')
ax.set_ylabel('Y')
plt.show()
```

```python
import matplotlib.pyplot as plt

x1, y1 = [1, 2, 3], [1.2, 1.8, 2.5]
x2, y2 = [1, 2, 3], [2.1, 2.4, 3.2]

fig, ax = plt.subplots()
ax.scatter(x1, y1, marker='o', label='Nhóm A')
ax.scatter(x2, y2, marker='s', label='Nhóm B')

ax.set_title('Scatter theo nhóm')
ax.set_xlabel('X'); ax.set_ylabel('Y')
ax.legend(title='Nhóm')
ax.grid(True, alpha=0.3)
plt.show()
```

```python

import seaborn as sns
import matplotlib.pyplot as plt

# Tải dữ liệu mẫu
tips = sns.load_dataset('tips')

fig, ax = plt.subplots()
sns.scatterplot(data=tips, x='total_bill', y='tip',
                hue='time', style='sex', ax=ax)

ax.set_title('Scatter Plot (Seaborn)')
ax.set_xlabel('Total bill ($)')
ax.set_ylabel('Tip ($)')

plt.show()
```
