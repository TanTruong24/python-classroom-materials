# Vẽ biểu đồ trong Python với matplotlib

Trong ví dụ này, chúng ta sẽ:
- Tạo dữ liệu mô phỏng theo công thức toán học
- Vẽ biểu đồ đường (line chart)
- Giải thích ý nghĩa của biểu đồ

Chúng ta sẽ dùng công thức hàm sin:

$$
y = \sin(x)
$$

Với \( x \) chạy từ 0 đến 2\pi.
.

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Tạo dữ liệu x từ 0 đến 2π, gồm 200 điểm
x = np.linspace(0, 2 * np.pi, 200)

# Áp dụng công thức y = sin(x)
y = np.sin(x)
```

```python
plt.figure(figsize=(8, 4))
plt.plot(x, y, linewidth=2)

plt.title("Biểu đồ hàm số y = sin(x)")
plt.xlabel("Giá trị x")
plt.ylabel("Giá trị y")

plt.grid(True)   # bật lưới cho dễ quan sát
plt.show()
```

### 🟦 Nhận xét biểu đồ

- Biểu đồ thể hiện dạng sóng hình sin quen thuộc.
- Hàm đạt giá trị lớn nhất **1** tại $( x = \frac{\pi}{2} $).
- Hàm đạt giá trị nhỏ nhất **-1** tại $( x = \frac{3\pi}{2} $).
- Đây là ví dụ điển hình của việc dùng **NumPy để tính toán** và **matplotlib để trực quan hóa**.
